### 2.3. Train-Test data split / Podział na dane treningowe i testowe

**Why split data into train and test sets before data cleaning and EDA?**
> 1. **Data leakage prevention:** Splitting raw data prior to missing value imputation, scaling, or encoding guarantees that dataset statistics (e.g., medians) are computed strictly on the training set.
> 2. **Mimicking production conditions:** The test set (`X_test`) represents entirely new, unseen customers, enabling an unbiased evaluation of the model's predictive performance.
> 3. **Stratification (`stratify=y`):** Ensures that the original class distribution (churn vs. non-churn) is preserved across both splits, eliminating the risk of random sampling bias.

**Dlaczego wykonujemy podział na dane treningowe i testowe przed czyszczeniem i EDA?**
> 1. **Ochrona przed wyciekiem danych (Data Leakage):** Podział surowych danych przed imputacją braków, skalowaniem czy kodowaniem gwarantuje, że statystyki (np. mediany) są obliczane wyłącznie na zbiorze treningowym.
> 2. **Odwzorowanie warunków produkcyjnych:** Zbiór testowy (`X_test`) reprezentuje całkowicie nowych, nieznałych wcześniej klientów, pozwalając na rzetelną ocenę zdolności predykcyjnych modelu.
> 3. **Stratyfikacja (`stratify=y`):** Zapewnia zachowanie identycznych proporcji klas (churn vs brak churnu) w obu zbiorach, eliminując ryzyko losowego skrzywienia próby.

```python
#defining features (X) and target variable (y) / określenie zmiennych objaśniających (X) i zmiennej objaśnianej (y)
X = raw_data.drop(columns=['Churn'])
y = raw_data['Churn']

#splitting raw data (80% train, 20% test) maintaining target class distribution (stratify) 
#podział surowych danych (80% train, 20% test) z zachowaniem proporcji Churnu (stratify)
raw_X_train, raw_X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.20, 
    random_state=42, 
    stratify=y
)

#combining X_train and y_train to create a raw dataset for cleaning and EDA
#połączenie X_train z y_train, aby utworzyć surowy zbiór do czyszczenia i analizy
raw_train_data = raw_X_train.copy()
raw_train_data['Churn'] = y_train

print(f"Number of rows in training set: {raw_train_data.shape[0]}")
print(f"Number of rows in test set (new customers): {raw_X_test.shape[0]}")
```

## 4. Data cleaning / Czyszczenie danych

### ENG

> 1. **Data cleaning objective:** Raw data contains duplicates, missing values (`NaN`), and outliers. Data cleaning ensures the stability of ML models, which cannot handle missing values directly.
> 
> 2. **Imputing missing values with median:** We use the **median** instead of the mean for numerical variables because the median is robust to outliers and skewed distributions.
> 
> 3. **Split (Train vs. Test):**
>    - **Training Set (`train_data`):** Imputation statistics (median) are calculated *strictly* based on the training data.
>    - **Test Set (`X_test`):** We apply the exact same medians calculated from the training set to impute missing values in the test set.
>    - **Why?** Calculating statistics on the full dataset would lead to **data leakage**. Separating the test set perfectly replicates production conditions, where the model receives new, raw customer data.

### PL

> 1. **Cel czyszczenia danych:** Surowe dane zawierają duplikaty, braki (`NaN`) oraz wartości odstające. Czyszczenie zapewnia stabilność modeli ML, które nie potrafią bezpośrednio przetwarzać brakujących wartości.
> 
> 2. **Uzupełnianie braków danych medianą:** Używamy **mediany** zamiast średniej dla zmiennych numerycznych, ponieważ mediana jest odporna na wartości skrajne (outliery) oraz skośność rozkładów.
>
> 3. **Podział (Trening vs. Test):**
>    - **Zbiór treningowy (`train_data`):** Statystyki do uzupełniania braków (mediana) wyliczamy *wyłącznie* na podstawie danych treningowych.
>    - **Zbiór testowy (`X_test`):** Dokładnie te same wyliczone wcześniej mediany z treningu aplikujemy do uzupełnienia braków w zbiorze testowym.
>    - **Dlaczego?** Wyliczenie statystyk na całym zbiorze doprowadziłoby do **wycieku danych (data leakage)**. Odseparowanie zbioru testowego idealnie odwzorowuje warunki produkcyjne, w których model przyjmuje nowe, surowe dane klientów.

### 4.4 Outliers / Wartości odstające

> 1. **Detection:** To identify extreme values, we utilize the Interquartile Range (IQR) method with bounds defined as $Q_1 - 1.5 \cdot \text{IQR}$ for the lower bound and $Q_3 + 1.5 \cdot \text{IQR}$ for the upper bound.
> 2. **Capping instead of removal:** To preserve valuable churn signals and avoid removing rare observations, extreme values are capped at the computed threshold boundaries rather than deleted.
> 3. **Data leakage prevention:** IQR boundaries are calculated **strictly on the training set (`train_data`)**. The exact same boundary values are subsequently applied to the test set (`X_test`) without recomputing the quartiles.

> 1. **Detekcja:** Do identyfikacji wartości skrajnych wykorzystujemy metodę zakresu międzykwartylowego (IQR) oraz regułę $Q_1 - 1.5 \cdot IQR$ dla dolnej granicy i $Q_3 + 1.5 \cdot IQR$ dla górnej granicy.
> 2. **Przycinanie (Capping) zamiast usuwania:** Aby zachować cenne sygnały dotyczące churnu i nie usuwać rzadkich obserwacji z bazy, wartości skrajne są przycinane do wyznaczonych wartości granicznych, a nie kasowane.
> 3. **Ochrona przed wyciekiem danych (Data Leakage):** Granice IQR są obliczane **wyłącznie na zbiorze treningowym (`train_data`)**. Następnie te same wartości graniczne są stosowane do zbioru testowego (`X_test`) bez ponownego przeliczania kwartyli.

```python
#selecting numerical columns for outliers detection / wybór kolumn numerycznych do sprawdzenia wartości odstających
num_cols = train_data.select_dtypes(include=[np.number]).drop(columns=['Churn', 'SatisfactionScore','NumberOfDeviceRegistered'], errors='ignore').columns

#boxplots for numerical variables before capping / boxploty dla zmiennych numerycznych przed przycięciem
plt.figure(figsize=(16, 8))
train_data[num_cols].boxplot(rot=45)
plt.title(
    'Distribution of Numerical Variables and Outliers (Before Capping)',
    fontsize=14,
)
plt.ylabel('Value')
plt.tight_layout()
plt.show()
```

```python
#dictionary to store bounds calculated EXCLUSIVELY on the training set / słownik do przechowywania granic wyliczonych WYŁĄCZNIE na zbiorze treningowym
iqr = {}

#calculating IQR bounds on the training set and cap values in train_data / obliczenie granic IQR na zbiorze treningowym i przycięcie wartości w train_data
for col in num_cols:
    Q1 = train_data[col].quantile(0.25)
    Q3 = train_data[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    #storing bounds derived from training data / zapisujemy granice z danych treningowych
    iqr[col] = (lower_bound, upper_bound)

    #identyfing outliers / identyfikacja outlierów
    outliers = train_data[
        (train_data[col] < lower_bound) | (train_data[col] > upper_bound)
    ]
    count = len(outliers)
    percentage = (count / len(train_data)) * 100

    print(
        f"Variable '{col}': {count} outliers ({percentage:.2f}% of total rows)"
    )
    #capping extreme values in the training set / przycinamy wartości w zbiorze treningowym
    train_data[col] = train_data[col].clip(lower=lower_bound, upper=upper_bound)

#applying the EXACT SAME training bounds to the test set (X_test) / aplikowanie TYCH SAMYCH granic ze zbioru treningowego do zbioru testowego (X_test)
for col in num_cols:
    if col in X_test.columns:
        lower_bound, upper_bound = iqr[col]
        X_test[col] = X_test[col].clip(lower=lower_bound, upper=upper_bound)
print("")
print("Capping successfully applied to both train_data and X_test using training bounds.")
```

#### 5.0.1. Descriptive statistics function / Funkcja statystyk opisowych

```python
#descriptive statistics for numerical variables / statystyki opisowe dla zmiennych numerycznych

def stats_summary(df):
    #only numeric / tylko kolumny numeryczne
    num_df = df.select_dtypes(include=[np.number])

    #statistics / obliczenie zbiorczych statystyk
    summary = pd.DataFrame(
        {
            'Count': num_df.count(),
            'Mean': num_df.mean().round(2),
            'Std Dev': num_df.std().round(2),
            'Median (Q2)': num_df.median(),
            'Mode': num_df.mode().iloc[0],  # pierwsza dominanta
            'IQR': (num_df.quantile(0.75) - num_df.quantile(0.25)).round(2),
            'Min': num_df.min(),
            'Max': num_df.max(),
            'Skewness': num_df.skew().round(2),
            'Kurtosis': num_df.kurt().round(2),
        }
    )

    return summary
```

#### 5.0.2. Barplot function / Funkcja wykresu słupkowego

```python
#function defining / definiowanie funkcji
def plot_distribution(
    df, column_name, title=None, palette='Blues_d', sort_by_index=True, order_by_count=False,
):
    #setting plot theme / ustawienie stylu wykresu
    sns.set_theme(style='whitegrid')
    plt.figure(figsize=(9, 5))

    #counting values / zliczanie wartości
    counts = df[column_name].value_counts()
    if sort_by_index:
        counts = counts.sort_index()

    counts_df = counts.reset_index()
    counts_df.columns = [column_name, 'Count']
    total_count = len(df[column_name].dropna())

    #ordering by count / sortowanie po zliczeniach
    if order_by_count:
        counts = counts.sort_values(ascending=False)  # Sortowanie malejąco
    elif sort_by_index:
        counts = counts.sort_index()

    counts_df = counts.reset_index()
    counts_df.columns = [column_name, 'Count']
    total_count = len(df[column_name].dropna())

    #barplot / wykres słupkowy
    ax = sns.barplot(
        data=counts_df,
        x=column_name,
        y='Count',
        palette=palette,
        order=counts_df[column_name],
        edgecolor='black',
        hue=column_name,
        legend=False,
    )

    #adding labels (values and %) / dodawanie etykiet (wartości i %) 
    for p in ax.patches:
        height = p.get_height()
        percentage = (height / total_count) * 100
        ax.annotate(
            f'{int(height)}\n({percentage:.1f}%)',
            (p.get_x() + p.get_width() / 2.0, height),
            ha='center',
            va='bottom',
            fontsize=10,
            fontweight='bold',
            xytext=(0, 3),
            textcoords='offset points',
        )

    #axis and title formatting / formatowanie osi i tytułu
    plot_title = (
        title
        if title
        else f'Distribution of {column_name.replace("_", " ").title()}'
    )
    plt.title(plot_title, fontsize=14, pad=15, weight='bold')
    plt.xlabel(column_name, fontsize=12, labelpad=10)
    plt.ylabel('Number of Customers', fontsize=12)
    plt.ylim(0, max(counts_df['Count']) * 1.18)
    sns.despine()

    plt.tight_layout()
    plt.show()
```

#### 5.0.3. Histplot function / Funkcja histogramu

```python
def plot_continuous_distribution(
    df, column_name, title=None, color='#2b5c8f', bins=30
):
    sns.set_theme(style='whitegrid')
    plt.figure(figsize=(9, 5))

    #histplot with KDE / histogram z krzywą gęstości KDE
    ax = sns.histplot(
        data=df,
        x=column_name,
        kde=True,
        color=color,
        bins=bins,
        edgecolor='black',
        alpha=0.6,
    )

    #imputing and adding lines for mean and median / obliczanie i dodanie linii dla średniej oraz mediany
    mean_val = df[column_name].mean()
    median_val = df[column_name].median()

    plt.axvline(
        mean_val,
        color='red',
        linestyle='--',
        linewidth=2,
        label=f'Mean: {mean_val:.2f}',
    )
    plt.axvline(
        median_val,
        color='green',
        linestyle='-',
        linewidth=2,
        label=f'Median: {median_val:.2f}',
    )

    #formatting axis and title / formatowanie osi i tytułu
    plot_title = (
        title
        if title
        else f'Distribution of {column_name.replace("_", " ").title()}'
    )
    plt.title(plot_title, fontsize=14, pad=15, weight='bold')
    plt.xlabel(column_name, fontsize=12, labelpad=10)
    plt.ylabel('Frequency', fontsize=12)
    plt.legend(fontsize=10)
    sns.despine()

    plt.tight_layout()
    plt.show()
```

### 5.2. Normality / Normalność

```python
normality_results = []

#computing normality statistics for each numeric feature / obliczanie statystyk normalności rozkładu dla każdej cechy numerycznej
for col in [c for c in num_cols if c != 'Complain']:
    #extracting series for the current column / wyciąganie serii danych dla bieżącej kolumny
    series = train_data[col]

    #calculating skewness and kurtosis / obliczanie skośności i kurtozy
    skewness = series.skew()
    kurt = series.kurtosis()

    #performing omnibus D'Agostino-Pearson test for normality / test D'Agostino-Pearsona na normalność
    stat, p_val = stats.normaltest(series)

    #determining asymmetry direction / określanie kierunku asymetrii
    if skewness > 0.5:
        skew_desc = 'right-skewed'
    elif skewness < -0.5:
        skew_desc = 'left-skewed'
    else:
        skew_desc = 'fairly symmetric'

    normality_results.append(
        {
            'Variable': col,
            'Skewness': round(skewness, 2),
            'Asymmetry': skew_desc,
            'Kurtosis': round(kurt, 2),
            'Omnibus stat': round(stat, 4),
            'p-value': f'{p_val:.4e}',
            'Is normal?': 'YES' if p_val >= 0.05 else 'NO (non-normal)',
        }
    )

#displaying the summary table / zbiorcza tabela 
df_normality = pd.DataFrame(normality_results)
display(df_normality)
```

```python
cols_to_plot = list(set(num_cols) - {'Complain'})
cols_per_row = 3
num_features = len(cols_to_plot)
num_rows = math.ceil(num_features / cols_per_row)

#creating a subplot grid / tworzenie siatki wykresów
fig, axes = plt.subplots(
    num_rows,
    cols_per_row,
    figsize=(5 * cols_per_row, 4 * num_rows),
    sharex=False,
    sharey=False,
)

axes_flat = axes.flatten() if num_features > 1 else [axes]

#loop for every feature / pętla dla każdej zmiennej
for i, feature in enumerate(cols_to_plot):
    ax = axes_flat[i]
    clean_series = train_data[feature].dropna()

    stats.probplot(clean_series, dist='norm', plot=ax)

    ax.set_title(f'Q-Q Plot: {feature}', fontsize=11, fontweight='bold')
    ax.set_xlabel('Kwantyle teoretyczne', fontsize=9)
    ax.set_ylabel('Kwantyle z próby', fontsize=9)
    ax.grid(True, linestyle='--', alpha=0.5)

#removing unused subplots / usunięcie nieużywanych kafelków w siatce
for j in range(num_features, len(axes_flat)):
    fig.delaxes(axes_flat[j])

plt.suptitle(
    'Q-Q normality plots',
    fontsize=14,
    fontweight='bold',
    y=1.02,
)
plt.tight_layout()
plt.show()
```

> **Normality testing (Omnibus D'Agostino-Pearson):**
> * For **all** evaluated numeric features, the $p$-value is drastically below the standard significance threshold $\alpha = 0.05$ ($p < 0.001$), leading to a definitive rejection of the null hypothesis of normality.
>   
> **Methodological implications:**
>   1. **Correlation Analysis:** replace Pearson’s linear correlation with **Spearman’s rank-order correlation** to handle non-Gaussian continuous/ordinal features.
>   2. **Hypothesis Testing:** use the non-parametric **U Mann-Whitney test** instead of the parametric Student’s $t$-test for group comparisons.
>   3. **Descriptive Aggregations:** maintain the **median** as the primary measures of central tendency across all reporting functions.

> **Test normalności rozkładów (Omnibus D'Agostino-Pearson):**
> * Dla **wszystkich** badanych zmiennych numerycznych wartość $p$ (*p-value*) jest drastycznie niższa od przyjętego poziomu istotności $\alpha = 0.05$ ($p < 0.001$). Oznacza to jednoznaczne odrzucenie hipotezy zerowej o normalności rozkładów, rozkład **nie jest normalny**.
> 
> **Konsekwencje:**
>   1. **Korelacje:** zamiast liniowego współczynnika korelacji Pearsona lepiej zastosować odporną na brak normalności **korelację rang Spearmana**.
>   2. **Porównywanie grup:** do porównywania zmiennych ciągłych między grupami zamiast testu $t$-Studenta lepiej użyć **nieparametrycznego testu U Manna-Whitneya**.
>   3. **Agregacje:** główną miarą tendencji centralnej w raportowaniu pozostaje **mediana** zamiast średniej arytmetycznej.

### 6.0. Plots & tests functions

#### 6.0.1. KDE Plot, U-Mann Whitney Test, Rank Biserial Correlation

```python
#defining kde plot & u-mann whitney test function / definiowanie funkcji wykresu kde i testu u-manna whitneya
def kde_test(
    data: pd.DataFrame,
    feature: str,
    target: str = "Churn",
    xlabel: str = None,
    alpha: float = 0.05,
    alternative: str = "two-sided",  # 'two-sided', 'less', lub 'greater'
    h0_text: str = None,  # Opcjonalny własny opis H0
    h1_text: str = None,  # Opcjonalny własny opis H1
    colors: list = None,
) -> None:
    if colors is None:
        colors = ["#2b5c8f", "#d95f02"]

    label_x = xlabel if xlabel is not None else feature

    #kde plot / wizualizacja wykresu gęstości (KDE)
    plt.figure(figsize=(10, 5))
    sns.kdeplot(
        data=data,
        x=feature,
        hue=target,
        common_norm=False,
        fill=True,
        palette=colors,
    )
    plt.title(f"Distribution of {feature} by {target}")
    plt.xlabel(label_x)
    plt.ylabel("Density")
    plt.tight_layout()
    plt.show()

    #preparing groups / przygotowanie grup
    group_0 = data[data[target] == 0][feature].dropna()
    group_1 = data[data[target] == 1][feature].dropna()

    #u-mann whitney test
    stat, p_value = stats.mannwhitneyu(
        group_0, group_1, alternative=alternative
    )

    #rank biserial correlation / siła efektu (korelacja dwurzędowa)
    r_rb = 1 - ((2 * stat) / (len(group_0) * len(group_1)))

    abs_r = abs(r_rb)
    if abs_r < 0.1:
        effect_desc = "negligible"
    elif abs_r < 0.3:
        effect_desc = "small"
    elif abs_r < 0.5:
        effect_desc = "medium"
    else:
        effect_desc = "large"

    #h0,h1
    if h0_text is None:
        h0_text = f"The distributions of {feature} are the same for group {target}=0 and group {target}=1."

    if h1_text is None:
        if alternative == "greater":
            h1_text = (
                f"Values of {feature} in the {target}=0 group are HIGHER than in {target}=1. "
                f"({feature} for churning clients is LOWER)."
            )
        elif alternative == "less":
            h1_text = (
                f"Values of {feature} in the {target}=0 group are LOWER than in {target}=1. "
                f"({feature} for churning clients is HIGHER)."
            )
        else:
            h1_text = f"The values of {feature} differ between group {target}=0 and group {target}=1."

    #results / wyniki
    print("")
    print(
        f"=== U Mann-Whitney test for: {feature} (alternative='{alternative}') ==="
    )
    print("")
    print(f"H0: {h0_text}")
    print(f"H1: {h1_text}")
    print("")
    print(f"U statistic : {stat:.2f}")
    print(f"p-value     : {p_value:.4e}")
    print("")

    if p_value < alpha:
        print(
            f"Decision    : Reject H0 (p < {alpha}). Difference is statistically significant."
        )
    else:
        print(
            f"Decision    : Failed to reject H0 (p >= {alpha}). Difference is not statistically significant."
        )

    if r_rb > 0:
        print(
            f"Direction   : {feature} values are HIGHER in the {target}=1 group."
        )
    else:
        print(
            f"Direction   : {feature} values are LOWER in the {target}=1 group."
        )
    print(f"Effect Size : r_rb = {r_rb:.2f} -> {effect_desc}")
```

```python
#defining u-mann whitney test function / definiowanie funkcji testu u-manna whitneya
def mannwhitney(
    data: pd.DataFrame,
    feature: str,
    target: str = "Churn",
    alpha: float = 0.05,
    alternative: str = "two-sided",  # 'two-sided', 'less', 'greater'
    h0_text: str = None,  
    h1_text: str = None):  

    #preparing groups / przygotowanie grup
    group_0 = data[data[target] == 0][feature].dropna()
    group_1 = data[data[target] == 1][feature].dropna()

    #U Mann-Whitney test 
    stat, p_value = stats.mannwhitneyu(
        group_0, group_1, alternative=alternative
    )

    #rank biserial correlation / siła efektu (korelacja dwurzędowa)
    r_rb = 1 - ((2 * stat) / (len(group_0) * len(group_1)))

    abs_r = abs(r_rb)
    if abs_r < 0.1:
        effect_desc = "negligible"
    elif abs_r < 0.3:
        effect_desc = "small"
    elif abs_r < 0.5:
        effect_desc = "medium"
    else:
        effect_desc = "large"

    # h0,h1
    if h0_text is None:
        h0_text = f"The distributions of {feature} are the same for group {target}=0 and group {target}=1."

    if h1_text is None:
        if alternative == "greater":
            h1_text = (
                f"Values of {feature} in the {target}=0 group are HIGHER than in {target}=1."
                f"({feature} for churning clients is LOWER)."
            )
        elif alternative == "less":
            h1_text = (
                f"Values of {feature} in the {target}=0 group are LOWER than in {target}=1. "
                f"({feature} for churning clients is HIGHER)."
            )
        else:
            h1_text = f"The values of {feature} differ between group {target}=0 and group {target}=1."

    #results / wyniki
    print("")
    print(
        f"=== U Mann-Whitney test for: {feature} (alternative='{alternative}') ==="
    )
    print("")
    print(f"H0: {h0_text}")
    print(f"H1: {h1_text}")
    print("")
    print(f"U statistic : {stat:.2f}")
    print(f"p-value     : {p_value:.4e}")
    print("")

    if p_value < alpha:
        print(
            f"Decision    : Reject H0 (p < {alpha}). Difference is statistically significant."
        )
    else:
        print(
            f"Decision    : Failed to reject H0 (p >= {alpha}). Difference is not statistically significant."
        )

    if r_rb > 0:
        print(
            f"Direction   : {feature} values are HIGHER in the {target}=1 group."
        )
    else:
        print(
            f"Direction   : {feature} values are LOWER in the {target}=1 group."
        )
    print(f"Effect Size : r_rb = {r_rb:.2f} -> {effect_desc}")
```

#### 6.0.2. Chi-Square test

```python
#defining chi2 test & plot / definiowanie funkcji testu chi2 i wykresu
def chi2(
    data: pd.DataFrame,
    feature: str,
    target: str = 'Churn',
    alpha: float = 0.05,
    colors: list = None,
    title: str = None,
    xlabel: str = None,
    h0_text: str = None,  # Opcjonalny własny opis H0
    h1_text: str = None,  # Opcjonalny własny opis H1
) -> None:
    if colors is None:
        colors = ['#2b5c8f', '#d95f02']

    label_x = xlabel if xlabel is not None else feature
    plot_title = (
        title
        if title is not None
        else f'{target} by {feature}'
    )

    #preparing data / przygotowywanie danych
    cross_tab = (
        pd.crosstab(
            train_data[feature], train_data[target], normalize='index'
        )
        * 100
    )

    #plot / wykres
    fig, ax = plt.subplots(figsize=(8, 5))
    cross_tab.plot(
        kind='bar',
        stacked=True,
        color=colors,
        ax=ax,
        width=0.4,
    )

    plt.title(plot_title, fontsize=12)
    plt.xlabel(label_x)
    plt.ylabel('(%)')
    plt.xticks(rotation=0)
    plt.ylim(0, 115)  
    plt.legend(title=target, labels=['0', '1'], loc='upper right')

    #percentages / procenty
    for p in ax.patches:
        height = p.get_height()
        if height > 0:
            ax.annotate(
                f'{height:.1f}%',
                (p.get_x() + p.get_width() / 2.0, p.get_y() + height / 2.0),
                ha='center',
                va='center',
                color='white',
                fontweight='bold',
                fontsize=10,
            )

    plt.tight_layout()
    plt.show()

    #test Chi-Square 
    contingency_table = pd.crosstab(train_data[feature], train_data[target])
    chi2, p_value, dof, expected = stats.chi2_contingency(contingency_table)

    #V Cramér 
    n = contingency_table.sum().sum()
    min_dim = min(contingency_table.shape) - 1
    cramers_v = np.sqrt(chi2 / (n * min_dim)) if min_dim > 0 else 0

    if cramers_v < 0.1:
        effect_desc = 'negligible'
    elif cramers_v < 0.3:
        effect_desc = 'small'
    elif cramers_v < 0.5:
        effect_desc = 'medium'
    else:
        effect_desc = 'large'

    #h0,h1
    if h0_text is None:
        h0_text = f"The distributions of {feature} are the same for group {target}=0 and group {target}=1."

    if h1_text is None:
         h1_text = f"The values of {feature} differ between group {target}=0 and group {target}=1."

    #results / wyniki
    print('')
    print(f'=== Chi-Square Test for: {feature} ===')
    print('')
    print(f"H0: {h0_text}")
    print(f"H1: {h1_text}")
    print('')
    print(f'Chi2 Statistic: {chi2:.2f}')
    print(f'p-value: {p_value:.4e}')
    print('')
    
    if p_value < alpha:
        print(
            f'The difference is statistically significant (rejecting H0) - {feature} has a significant impact on {target}.'
        )
    else:
        print(
            f'Failed to reject H0 - the difference for {feature} is not statistically significant.'
        )
    print(f"Cramer's V: {cramers_v:.2f} -> {effect_desc}")
```
